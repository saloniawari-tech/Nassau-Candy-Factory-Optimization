```python
# LOAD OPTIMIZATION OUTPUTS FOR STREAMLIT

import pandas as pd
import numpy as np

recommendations = pd.read_csv(
    "../data/processed/factory_recommendations.csv"
)

scenarios = pd.read_csv(
    "../data/processed/factory_scenarios.csv"
)

kpis = pd.read_csv(
    "../data/processed/optimization_kpis.csv"
)

print("Dashboard datasets loaded successfully!")

print("\nRecommendations:", recommendations.shape)
print("Scenarios:", scenarios.shape)
print("KPIs:", kpis.shape)
```

```python
# INSPECT DASHBOARD DATA

print("RECOMMENDATION DATA")
print("=" * 40)

print("\nColumns:")
print(recommendations.columns.tolist())

print("\nRecommendation counts:")
print(
    recommendations["Recommendation"]
    .value_counts()
)

print("\nKPI DATA:")
display(kpis)

print("\nSample recommendations:")
display(recommendations.head(5))
```

```python
# PREPARE STREAMLIT KPI VALUES

kpi_values = {
    row["KPI"]: row["Value"]
    for _, row in kpis.iterrows()
}

avg_lead_reduction = kpi_values[
    "Average Lead Time Reduction"
]

profit_stability = kpi_values[
    "Profit Impact Stability"
]

scenario_confidence = kpi_values[
    "Scenario Confidence Score"
]

recommendation_coverage = kpi_values[
    "Recommendation Coverage"
]

print("Streamlit KPI values prepared!")

print(f"Average Lead Time Reduction: {avg_lead_reduction:.2f}%")
print(f"Profit Impact Stability: {profit_stability:.2f}%")
print(f"Scenario Confidence: {scenario_confidence:.2f}%")
print(f"Recommendation Coverage: {recommendation_coverage:.2f}%")
```

```python
# PREPARE STREAMLIT FILTER OPTIONS

products = sorted(
    recommendations["Product Name"].unique()
)

regions = sorted(
    recommendations["Region"].unique()
)

factories = sorted(
    scenarios["Scenario Factory"].unique()
)

print("Streamlit filter options prepared!")

print("\nProducts:", len(products))
print(products)

print("\nRegions:", regions)

print("\nFactories:", factories)
```

```python
# PREPARE WHAT-IF SCENARIO DATA

what_if_data = scenarios[
    [
        "Product Name",
        "Region",
        "Current Factory",
        "Scenario Factory",
        "Current Avg Lead Time",
        "Predicted Lead Time",
        "Lead Time Improvement",
        "Lead Time Reduction %",
        "Distance km",
        "Profit Margin %",
        "Profit Impact Stable",
        "Recommendation Score"
    ]
].copy()

what_if_data = what_if_data.sort_values(
    [
        "Product Name",
        "Region",
        "Predicted Lead Time"
    ]
).reset_index(drop=True)

print("What-if scenario data prepared!")
print("Rows:", len(what_if_data))

display(what_if_data.head(10))
```

```python
# PREPARE RECOMMENDATION DASHBOARD DATA

recommendation_dashboard = recommendations[
    [
        "Product Name",
        "Region",
        "Current Factory",
        "Scenario Factory",
        "Lead Time Reduction %",
        "Distance km",
        "Profit Stability Score",
        "Recommendation Score",
        "Recommendation"
    ]
].copy()

# Add a simple status for dashboard display
recommendation_dashboard["Status"] = np.where(
    recommendation_dashboard["Recommendation"]
    == "Recommend Reassignment",
    "Reassignment Recommended",
    "Current Factory Retained"
)

print("Recommendation dashboard data prepared!")
print("Rows:", len(recommendation_dashboard))

display(recommendation_dashboard.head(10))
```

```python
# PREPARE RISK & IMPACT DATA

risk_impact = scenarios[
    [
        "Product Name",
        "Region",
        "Current Factory",
        "Scenario Factory",
        "Lead Time Reduction %",
        "Profit Margin %",
        "Profit Impact Stable",
        "Recommendation Score"
    ]
].copy()

# Risk classification
risk_impact["Risk Level"] = np.select(
    [
        risk_impact["Lead Time Reduction %"] < -10,
        risk_impact["Lead Time Reduction %"] < 0
    ],
    [
        "High Risk",
        "Moderate Risk"
    ],
    default="Low Risk"
)

print("Risk & Impact data prepared!")

print("\nRisk distribution:")
print(
    risk_impact["Risk Level"]
    .value_counts()
)

display(risk_impact.head(10))
```

```python
# PREPARE SHIP MODE OPTIONS

raw_df = pd.read_csv("../data/Nassau Candy Distributor.csv")

ship_modes = sorted(
    raw_df["Ship Mode"].dropna().unique()
)

print("Ship Mode options prepared!")
print("Ship Modes:", ship_modes)
```

```python
# SAVE DASHBOARD-READY DATA

dashboard_path = "../data/processed/"

what_if_data.to_csv(
    dashboard_path + "dashboard_what_if.csv",
    index=False
)

recommendation_dashboard.to_csv(
    dashboard_path + "dashboard_recommendations.csv",
    index=False
)

risk_impact.to_csv(
    dashboard_path + "dashboard_risk_impact.csv",
    index=False
)

print("Dashboard-ready datasets saved successfully!")

print("\nFiles created:")
print("- dashboard_what_if.csv")
print("- dashboard_recommendations.csv")
print("- dashboard_risk_impact.csv")
```
